# Homework 1: Stock Price Analysis
## Group 3: Stocks starting with K, L, M

Download 10 stocks from Yahoo Finance (US stock market) with letters K, L, M
- Date range: 01-01-2015 to 31-07-2025
- Data: Date and Closing price only

```python
# Import required libraries
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

## A & B: Download Stock Data and Create DataFrame

Selected stocks:
- **K**: KO (Coca-Cola), KHC (Kraft Heinz), KEY (KeyCorp)
- **L**: LMT (Lockheed Martin), LOW (Lowe's), LUV (Southwest Airlines), LLY (Eli Lilly)
- **M**: MCD (McDonald's), MSFT (Microsoft), META (Meta Platforms)

```python
# Define stock tickers for Group 3 (K, L, M)
tickers = ['KO', 'KHC', 'KEY', 'LMT', 'LOW', 'LUV', 'LLY', 'MCD', 'MSFT', 'META']

# Define date range
start_date = '2015-01-01'
end_date = '2025-07-31'

# Download data for all stocks
print(f"Downloading data for {len(tickers)} stocks from {start_date} to {end_date}...")
data = yf.download(tickers, start=start_date, end=end_date, progress=True)
```

```python
# Extract only the Closing prices
df = data['Close'].copy()

# Check the shape of the dataframe
print(f"\nDataFrame shape: {df.shape}")
print(f"Number of trading days: {len(df)}")
print(f"Number of stocks: {len(df.columns)}")
```

## C: Display DataFrame Header

```python
# Show the header of the dataframe
print("First few rows of the DataFrame:")
df.head()
```

```python
# Additional information about the data
print("\nDataFrame Info:")
df.info()
```

```python
# Check for any missing values
print("\nMissing values per stock:")
df.isnull().sum()
```

## D: Plot Daily Closing Prices

```python
# Create a comprehensive plot showing all stock prices over time
plt.figure(figsize=(16, 10))

# Plot each stock
for column in df.columns:
    plt.plot(df.index, df[column], label=column, linewidth=1.5)

# Customize the plot
plt.title('Daily Closing Prices - Group 3 Stocks (K, L, M)\n2015-01-01 to 2025-07-31', 
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Date', fontsize=12, fontweight='bold')
plt.ylabel('Closing Price (USD)', fontsize=12, fontweight='bold')
plt.legend(loc='best', fontsize=10, ncol=2)
plt.grid(True, alpha=0.3, linestyle='--')
plt.xticks(rotation=45)
plt.tight_layout()

# Display the plot
plt.show()
```

## Additional Analysis: Individual Stock Subplots

```python
# Create subplots for better individual stock visibility
fig, axes = plt.subplots(5, 2, figsize=(16, 20))
fig.suptitle('Individual Stock Price Variations (2015-2025)', fontsize=18, fontweight='bold', y=0.995)

# Flatten axes array for easier iteration
axes = axes.flatten()

# Plot each stock in its own subplot
for idx, column in enumerate(df.columns):
    axes[idx].plot(df.index, df[column], linewidth=1.5, color=f'C{idx}')
    axes[idx].set_title(f'{column}', fontsize=12, fontweight='bold')
    axes[idx].set_xlabel('Date', fontsize=10)
    axes[idx].set_ylabel('Price (USD)', fontsize=10)
    axes[idx].grid(True, alpha=0.3, linestyle='--')
    axes[idx].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```
